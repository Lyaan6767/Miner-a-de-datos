## Minado de patrones frecuentes

Ejemplo: Youtube registra las reproducciones de los usuarios en una sesión de visualización. Con millones de esos registros, puede hacer recomendaciones

Cada interacción entre el usuario/cliente con el proveedor/vendedor se llama transacción

## Notación


Nuestro catálogo de productios se denota por $$I = \{i_1, i_2, \dots, i_m\}$$ (la notación es por el inglés *Items*)

Cada transacción es un subconjunto de nuestro catálogo de items, es decir $$t \subseteq I$$

Nuestro registro de producto de interacciones entre usuarios y proveedores es una *base de datros transaccional*, representada por
$$D = \{ t_1, t_2, \dots, t_n \} \qquad \text{donde cada transacción cumple } t_k \subseteq I$$

###  Ejemplo

| Transacción | Artículos |
| --- | --- |
| $t_1$ | laptop, cámara, memoria SD |
| $t_2$ | laptop, mouse, memoria SD |
| $t_3$ | cámara, memoria SD, tripié |
| $t_4$ | laptop, cámara, mochila |


### Una medida: support

$$\mathrm{support}(X) = \frac{|\{t \in D: X \subseteq t\}|}{|D|}$$

¿Cuál es el *support* de $X_1= \{\text{laptop, memoria SD}\}$?


$$\mathrm{support}(X_1) = \frac{|\{t_1, t_2 \}|}{|D|}= \frac{2}{4}=0.5$$

¿Cuál es el *support* de $X_2= \{\text{memoria SD}\}$?
$$\mathrm{support}(X_2) = \frac{|\{t_1, t_2, t_3 \}|}{|D|}= \frac{3}{4}=0.75$$

### Reglas de asociación

Una **regla de asociación** es una implicación entre dos *itemsets* disjuntos:
$$X \Rightarrow Y, \qquad X \cap Y = \emptyset$$
 
Se lee: "quien lleva $X$ tiende a llevar también $Y$"

¿Cómo evaluamos qué tan buena es la regla?

#usaremos la confianza para evaluar qué tan buena es

$$\mathrm{conf}(X \Rightarrow Y) = \frac{\mathrm{support}(X \cup Y)}{\mathrm{support}(X)}$$

Esto es similar a 
$$P(Y | X) = \frac{P(X \cap Y)}{P(X)} $$

Otra medida para evaluar las reglas de asociación es la siguiente:

$$ \mathrm{lift}(X \Rightarrow Y) =
    \frac{\mathrm{conf}(X \Rightarrow Y)}
        {\mathrm{support}(Y)}
$$

- $\mathrm{lift} > 1$: llevar $X$ **aumenta** la propensión a llevar $Y$.
- $\mathrm{lift} = 1$: $X$ y $Y$ son independientes; la regla no aporta información.
- $\mathrm{lift} < 1$: llevar $X$ **reduce** la propensión a llevar $Y$.

Regla 2: $\{mochila\} \Rightarrow \{laptop\}$
$\mathrm{conf}(\{mochila \} \Rightarrow \{laptop\}) = \frac{\mathrm{support}(\{mochila, laptop\})}{\mathrm{support}(\{mochila\})}$


```python
support({'mochila', 'laptop'}, D) / support({'mochila'}, D)
```

```python
def conf(X, Y, D):
    x_union_y = X.union(Y)
    ans = support(x_union_y, D) / support(X, D)
    return ans
```

```python
conf({'laptop'}, {'cámara'}, D)
```

```python
conf({'mochila'}, {'laptop'}, D)
```

```python
def lift(X, Y, D):
    return conf(X, Y, D) / support(Y, D)
```

### ¿Cómo encontrar itemsets frecuentes?

La frecuencia se define de forma arbitraria. Si mi frecuencia mínima es $min\_ support$, los itemsets frecuentes son todos los $$X \subseteq I: \mathrm{support}(X) \geq \mathrm{min\_ support} $$

Una estrategia simple es utilizar fuerza bruta, es decir, probar los $2^{|I|}$ subconjuntos


```python
from itertools import combinations
```

```python
def itemsets_frequentes(D, min_support):
    """Todos los X subconjunto de I, con support(X) >= min_support"""
    I = sorted(set().union(*D), key=clave)
    encontrados = []
    # Tamaño de los itemsets:1 <= |X| <= len(I), es decir, |I|
    for k in range(1, len(I)+1):                        # Explora todos los subconjuntos de tamaño 1 hasta |I|
        for candidato in combinations(I, k):               #Explora todas las combinaciones
            sup = support(set(candidato), D)
            if sup >= min_support:
                encontrados.append((frozenset(candidato), sup))
    return encontrados
```

```python
import pandas as pd
```

```python
def generar_reglas(D, min_support, min_confidence):
    """Para cada itemset frecuente Z, prueba toda partición Z = X ∪ Y, con X, Y no vacíos"""
    reglas = []
    for Z, sup_Z in itemsets_frequentes(D, min_support):
        if len(Z) < 2:
            continue
    for r in range(1, len(Z)):
        for X in combinations(sorted(Z, key=clave), r):
            X, Y = frozenset(X), Z - frozenset(X)
            c = conf(X, Y, D)
            if c >= min_confidence:
                reglas.append({
                    'regla': f'{{{", ".join(sorted(X, key=clave))}}} then'
                            f'{{{", ".join(sorted(X, key=clave))}}}',
                    'support': round(sup_Z, 3),
                    'conf': round(c, 3),
                    'lift': round(lift(X, Y, D), 3)
                })
    return pd.DataFrame(reglas).sort_values('lift', ascending=False, ignore_index=True)
```

```python
generar_reglas(D, min_support=0.5, min_confidence=0.6)
```
