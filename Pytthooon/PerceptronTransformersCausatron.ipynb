```python
w1_0 = 0
w2_0 = 0
b_0 = -0.5
eta = 1

def y_hat(x):
    if x >=0:
        return 1
    else:
        return 0
```

```python
w1 = w1_0
w2 = w2_0
b = b_0
Contador = 0
```

```python
epoch_counter = 0
while(Contador != dfXD.shape[0]):
    epoch_counter+=1
    print(f'Epoch {epoch_counter}')
    for i in range(dfXD.shape[0]):
        print(f'Ejemplo {i}')
        x1 = dfXD.loc[i]['x1'] 
        x2 = dfXD.loc[i]['x2']
        y = df2.loc[i]['or']

        z = w1* x1 + w2* x2 + b
        y_calc = y_hat(z)
        error = y - y_calc
        if error == 0:
            Contador+=1
        else:
            Contador=0
        print(f'w=({w1}, {w2}), b={b}')
        w1 = w2 + eta * error * x1
        w2 = w2 + eta * error * x2
        b = b + eta * error
        print(f'x:({x1}, {x2}), y={y}, z={z}, y_hat={y_calc}, new_w=({w1}, {w2}), new_b={b}')
        if Contador == dfXD.shape[0]:
            break
    print('-------------------')
```
