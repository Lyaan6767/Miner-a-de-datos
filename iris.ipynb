```python
iris_df = pd.DataFrame(data=iris_data['data'], columns=iris_data['feature_names'])
```

```python
iris_df['Iris type'] = iris_data['target']
```

```python
iris_df.head()
```

```python
iris_df['Iris name'] = iris_df['Iris type'].apply(lambda x: 'setosa' if x == 0 else ('versicolor' if x == 1 else 'virginica'))
```

```python
sns.scatterplot(data= iris_df, x= "sepal length (cm)", y= "sepal width (cm)", hue= 'Iris name')
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
                                                    iris_df[["sepal length (cm)", "sepal width (cm)","petal length (cm)", "petal width (cm)"]],
                                                    iris_df[["Iris name"]],
                                                    test_size=0.15,
                                                    random_state=0
)
```

```python
from sklearn import preprocessing
```

```python
X_train_norm = preprocessing.normalize(X_train)
X_test_norm = preprocessing.normalize(X_test)
```

```python
type(X_train_norm)
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=0, n_init='auto')
kmeans.fit(X_train_norm)
```

```python
sns.scatterplot(data=X_train, x= 'sepal length (cm)', y='sepal width (cm)', hue=kmeans.labels_)
```

```python
sns.scatterplot(data=X_train, x= 'petal length (cm)', y='sepal width (cm)', hue=kmeans.labels_)
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_score(X_train_norm, kmeans.labels_, metric='euclidean')
```
