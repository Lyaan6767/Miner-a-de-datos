```python
def print_confusion_matrix(confusion_matrix, class_names, figsize=(9, 7), fontsize=14):
    df_cm = pd.DataFrame(confusion_matrix, index=class_names, columns=class_names)
    fig = plt.figure(figsize=figsize)
    try: 
        heatmap = sns.heatmap(df_cm, annot=True, fmt='d', cbar=True)
    except ValueError:
        raise ValueError('Confusion matrix values must be integers')
    heatmap.yaxis.set_ticklabels(heatmap.yaxis.get_ticklabels(), rotation=0, ha='right', fontsize=fontsize)
    heatmap.xaxis.set_ticklabels(heatmap.xaxis.get_ticklabels(), rotation=45, ha='right', fontsize=fontsize)
    plt.ylabel('True label', fontsize=12)
    plt.xlabel('Predicted label', fontsize=12)
    plt.title('Confusion Matrix', fontsize=16)
```

```python
def euclidean_distance(v1, v2):
    v1= v1.values[:4]
    v2= v2.values[:4]
    distance = 0
    for i in range(len(v1)):
        distance += (v1[i] - v2[i]) ** 2
    return np.sqrt(distance)
```

```python
def predict(k, train_set, train_labels, test_instance, y_test):
    distances= []
    for i in range(len(train_set)):
        dist = euclidean_distance(train_set.iloc[i], test_instance)
        distances.append((dist, train_labels.iloc[i]))
    y_pred = pd.DataFrame(data=distances, columns=['distancia', 'clase']).sort_values(by='distancia').head(k)['clase'].mode()[0]
    true_y = y_test
    ans = {'y_pred':y_pred, 'true_y':true_y}
    return ans
```

```python
def k_fold_indices(n, k, shuffle=True, seed=0): ## n tamaño de nuestro dataset, k bloques en los queremos separar, shuffle true es para reacomodarlos, false si queremos que se queden igual, seed=0 para...
    idx = np.arange(n)
    if shuffle: ##si verdadero, reordenamos
        np.random.default_rng(seed).shuffle(idx)
    folds = np.array_split(idx, k) ##aquí lo partimos en k partes al arreglo k
    for i in range(k):
        val =folds[i]
        train = np.concatenate([folds[j] for j in range(k) if j!=i])
        yield train, val ##te da una estructura en la que puedes seguir recorriendo y te la lo que tú le pongas, eso hace yield
```

```python
accu_for_k = {}
for num_neighbors in range (3, 30, 2):
    accu_list = []
    for train_set, val_set in k_fold_indices(len(X), 5, seed=45):
        
        KNN = KNeighborsClassifier(n_neighbors=num_neighbors)
        X_train = X.iloc[train_set]
        y_train = y.iloc[train_set]
        KNN.fit(X_train, y_train) #el fit para entrenarlo
        X_val = X.iloc[val_set]
        y_val = y.iloc[val_set]
        trained_copy = copy.deepcopy(KNN)
        val_evaluation = pd.DataFrame()
        val_evaluation['predicted'] = trained_copy.predict(X_val)
        val_evaluation['real'] = y_val.reset_index(drop=True)
        val_evaluation['is_correct'] = val_evaluation['predicted'] == val_evaluation['real']
        accuracy = val_evaluation['is_correct'].sum() / val_evaluation.shape[0]
        #print(f'La precisión obtenida es: {accuracy}')
        accu_list.append(accuracy)
    accu_for_k [num_neighbors]= sum(accu_list) / len(accu_list)
        
#print(sum(accu_list)/len(accu_list))
```

```python
max(accu_for_k, key=accu_for_k.get) #del diccionario que acabamos de crear, obtenemos el máximo y lo guardamos en key, el valor mostrado
                                     #cambia dependiendo de la semilla, si seed=45 entonces max = 15, si seed=21 entonces max da 11
```
